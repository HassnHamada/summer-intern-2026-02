```python
import matplotlib.pyplot as plt
import pandas as pd

def create_scatter_plot(
    x_column,
    y_column,
    number_of_rows=10
):
    # Check that both columns exist
    if x_column not in df.columns:
        return f"{x_column} was not found."

    if y_column not in df.columns:
        return f"{y_column} was not found."

    # Select only the requested rows
    selected_data = df.head(number_of_rows).copy()

    # Keep only X and Y columns
    plot_data = selected_data[[x_column, y_column]].copy()

    # Convert the values to numbers
    plot_data[x_column] = pd.to_numeric(
        plot_data[x_column],
        errors="coerce"
    )

    plot_data[y_column] = pd.to_numeric(
        plot_data[y_column],
        errors="coerce"
    )

    # Remove invalid rows
    plot_data = plot_data.dropna()

    print("Rows used:")
    display(selected_data)

    # Create the scatter plot
    plt.figure(figsize=(9, 5))

    plt.scatter(
        plot_data[x_column],
        plot_data[y_column]
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(f"{y_column} vs {x_column}")
    plt.grid(True)
    plt.show()

    return "Scatter plot created successfully."
```

```python
import pandas as pd
import matplotlib.pyplot as plt


def create_scatter_plot(
    number_of_rows: int = 10,
    x_column: str = "id",
    y_column: str = "history_score"
) -> str:

    # Convert row count into an integer
    number_of_rows = int(number_of_rows)

    # Check the columns
    if x_column not in df.columns:
        return (
            f"Column '{x_column}' does not exist. "
            f"Available columns: {df.columns.tolist()}"
        )

    if y_column not in df.columns:
        return (
            f"Column '{y_column}' does not exist. "
            f"Available columns: {df.columns.tolist()}"
        )

    # Take only the requested rows
    selected_rows = df.head(number_of_rows).copy()

    # Take the two columns used for plotting
    plot_data = selected_rows[[x_column, y_column]].copy()

    # Convert both axes into numeric values
    plot_data[x_column] = pd.to_numeric(
        plot_data[x_column],
        errors="coerce"
    )

    plot_data[y_column] = pd.to_numeric(
        plot_data[y_column],
        errors="coerce"
    )

    plot_data = plot_data.dropna()

    # Display the requested rows
    print(f"First {number_of_rows} rows:")
    display(selected_rows)

    # Create the scatter plot
    plt.figure(figsize=(9, 5))

    plt.scatter(
        plot_data[x_column],
        plot_data[y_column]
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(f"{y_column} vs {x_column}")

    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return (
        f"Displayed the first {number_of_rows} rows and created "
        f"a scatter plot with {x_column} as X and {y_column} as Y."
    )
```

```python
import ollama

messages = [
    {
        "role": "system",
        "content": f"""
You are a CSV visualization assistant.

Available columns:
{df.columns.tolist()}

When the user requests rows or a scatter plot, call
create_scatter_plot.

Rules:
- If the user only specifies a number of rows, use:
  x_column = "id"
  y_column = "history_score"
- Default number of rows is 10.
- Use the exact dataframe column names.
- If the user specifies different X and Y axes, use them.
"""
    }
]
```

```python
def ask_agent(question: str) -> str:

    messages.append({
        "role": "user",
        "content": question
    })

    response = ollama.chat(
        model="llama3.2:3b",
        messages=messages,
        tools=[create_scatter_plot]
    )

    assistant_message = response["message"]
    messages.append(assistant_message)

    tool_calls = assistant_message.get("tool_calls", [])

    if not tool_calls:
        return assistant_message.get("content", "No answer returned.")

    for tool_call in tool_calls:

        function_name = tool_call["function"]["name"]
        arguments = tool_call["function"]["arguments"]

        if function_name == "create_scatter_plot":
            result = create_scatter_plot(**arguments)
        else:
            result = f"Unknown function: {function_name}"

        messages.append({
            "role": "tool",
            "tool_name": function_name,
            "content": str(result)
        })

    final_response = ollama.chat(
        model="llama3.2:3b",
        messages=messages,
        tools=[create_scatter_plot]
    )

    final_answer = final_response["message"].get(
        "content",
        str(result)
    )

    messages.append(final_response["message"])

    return final_answer
```

```python
def create_scatter_plot(
    number_of_rows: int = 10,
    x_column: str = "id",
    y_column: str = "history_score"
) -> str:
    """
    Display a requested number of rows from the student CSV file
    and create a scatter plot from two selected numeric columns.

    Use this tool when the user asks to:
    - show the first N students or rows;
    - create a scatter plot;
    - select an X-axis and Y-axis;
    - plot one student score against another column.

    Args:
        number_of_rows:
            Number of rows to take from the beginning of the dataframe.
            Example: 3 means use the first 3 rows.

        x_column:
            Exact dataframe column name to use for the X-axis.
            Default is "id".

        y_column:
            Exact dataframe column name to use for the Y-axis.
            Default is "history_score".

    Returns:
        A text message confirming which rows and axes were used.
    """

    number_of_rows = int(number_of_rows)

    if x_column not in df.columns:
        return (
            f"Column '{x_column}' does not exist. "
            f"Available columns: {df.columns.tolist()}"
        )

    if y_column not in df.columns:
        return (
            f"Column '{y_column}' does not exist. "
            f"Available columns: {df.columns.tolist()}"
        )

    selected_rows = df.head(number_of_rows).copy()
    plot_data = selected_rows[[x_column, y_column]].copy()

    plot_data[x_column] = pd.to_numeric(
        plot_data[x_column],
        errors="coerce"
    )

    plot_data[y_column] = pd.to_numeric(
        plot_data[y_column],
        errors="coerce"
    )

    plot_data = plot_data.dropna()

    print(f"First {number_of_rows} rows:")
    display(selected_rows)

    plt.figure(figsize=(9, 5))
    plt.scatter(
        plot_data[x_column],
        plot_data[y_column]
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(f"{y_column} vs {x_column}")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return (
        f"Displayed the first {number_of_rows} rows and created "
        f"a scatter plot using '{x_column}' as X and "
        f"'{y_column}' as Y."
    )
```
