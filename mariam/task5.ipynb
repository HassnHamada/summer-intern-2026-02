# Neural Network 

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import tensorflow_datasets as tfds
```

```python
dataset, info = tfds.load(
    "oxford_flowers102",
    as_supervised=True,
    with_info=True
)
```

```python
IMG_SIZE = 128
def preprocess(image, label):
    image = tf.image.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    image = image / 255.0

    return image, label
```

```python
train_data = dataset["train"].map(preprocess)
val_data = dataset["validation"].map(preprocess)
test_data = dataset["test"].map(preprocess)
```

```python
model = keras.Sequential([
    
    layers.Flatten(
        input_shape=(128,128,3)
    ),

    layers.Dense(
        256,
        activation="relu"
    ),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dense(
        102,
        activation="softmax"
    )
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
BATCH_SIZE = 32

train_data = train_data.batch(BATCH_SIZE)
val_data = val_data.batch(BATCH_SIZE)
test_data = test_data.batch(BATCH_SIZE)
```

```python
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=10
)
```

```python
test_loss, test_accuracy = model.evaluate(test_data)

print("Test accuracy:", test_accuracy)
```

```python
import numpy as np

for image, label in test_data.take(1):

    prediction = model.predict(image)

    predicted_class = np.argmax(prediction[0])

    print("Predicted:", predicted_class)
    print("Actual:", label[0].numpy())
```

```python
np.argmax(prediction[0])
```

# Transfer Learning

```python
dataset2, info = tfds.load(
    "oxford_flowers102",
    as_supervised=True,
    with_info=True
)
```

```python
from tensorflow.keras.applications import MobileNetV2
```

```python
base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224,224,3)
)
```

```python
base_model.trainable = False
```

```python
import tensorflow as tf

IMG_SIZE = 224

def preprocess(image, label):
    image = tf.image.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    image = tf.keras.applications.mobilenet_v2.preprocess_input(image)

    return image, label
```

```python
from tensorflow.keras import layers, models


model_tl = models.Sequential([

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        102,
        activation="softmax"
    )
])
```

```python
model_tl.summary()
```

```python
train_data = dataset2["train"].map(preprocess)
val_data = dataset2["validation"].map(preprocess)
test_data = dataset2["test"].map(preprocess)
```

```python
BATCH_SIZE = 32

train_data = train_data.batch(BATCH_SIZE)
val_data = val_data.batch(BATCH_SIZE)
test_data = test_data.batch(BATCH_SIZE)
```

```python
model_tl.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
len(train_data)
```

```python
history_tl = model_tl.fit(
    test_data,
    validation_data=val_data,
    epochs=10
)
```

```python
train_loss, train_acc = model_tl.evaluate(train_data)

print(train_loss, train_acc)
```
