# Neural Network 

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
model = keras.Sequential([
    
    layers.Flatten(input_shape=(28,28)),
    
    layers.Dense(128, activation='relu'),
    
    layers.Dense(64, activation='relu'),
    
    layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)
```

```python
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test
)

print(test_accuracy)
```

```python
import numpy as np

prediction = model.predict(
    x_test[0].reshape(1,28,28)
)
```

```python
np.argmax(prediction)
```

# CNN

```python
(m_train, n_train), (m_test, n_test) = keras.datasets.mnist.load_data()
```

```python
m_train.shape
```

```python
m_train = m_train / 255.0
m_test = m_test / 255.0
```

```python
m_train = m_train.reshape(-1, 28, 28, 1)

m_test = m_test.reshape(-1, 28, 28, 1)
```

```python
m_train.shape
```

```python
model = keras.Sequential([
    
    layers.Conv2D(
        32,
        (3,3),
        activation='relu',
        input_shape=(28,28,1)
    ),

    layers.MaxPooling2D((2,2)),


    layers.Conv2D(
        64,
        (3,3),
        activation='relu'
    ),

    layers.MaxPooling2D((2,2)),


    layers.Flatten(),


    layers.Dense(
        128,
        activation='relu'
    ),


    layers.Dense(
        10,
        activation='softmax'
    )
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history2 = model.fit(
    m_train,
    n_train,
    epochs=5,
    validation_split=0.1
)
```

```python
test_loss, test_accuracy = model.evaluate(
    m_test,
    n_test
)

print(test_accuracy)
```

```python
import numpy as np

prediction = model.predict(
    m_test[0].reshape(1,28,28)
)
```

```python
np.argmax(prediction)
```
