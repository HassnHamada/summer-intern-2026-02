```python
from resemblyzer import VoiceEncoder, preprocess_wav
import librosa
import soundfile as sf
import numpy as np
import os

encoder = VoiceEncoder()

audio = preprocess_wav("audio.wav")

print("Encoder Loaded")
```

```python
sr = 16000

chunk_duration = 2

samples = sr * chunk_duration

os.makedirs("chunks", exist_ok=True)

paths = []

for i in range(0, len(audio), samples):

    chunk = audio[i:i+samples]

    if len(chunk) < sr:
        continue

    filename = f"chunks/chunk_{len(paths)}.wav"

    sf.write(filename, chunk, sr)

    paths.append(filename)

print(len(paths))
```

```python
embeddings = []

for path in paths:

    wav = preprocess_wav(path)

    embeddings.append(
        encoder.embed_utterance(wav)
    )

embeddings = np.array(embeddings)

print(embeddings.shape)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=2,
    random_state=42
)

labels = kmeans.fit_predict(embeddings)

print(labels)
```

```python
for i, label in enumerate(labels):

    start = i * 2
    end = start + 2

    print(
        f"{start:02d}s --> {end:02d}s : Speaker {label+1}"
    )
```

```python
from faster_whisper import WhisperModel

model = WhisperModel("small", device="cpu")

segments, _ = model.transcribe("audio.wav")

for seg in segments:

    idx = int(seg.start // 2)

    if idx >= len(labels):
        idx = len(labels) - 1

    print(
        f"Speaker {labels[idx]+1}: {seg.text}"
    )
```
