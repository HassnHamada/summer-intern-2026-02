```python
def chat_with_memory(user_input):

    # Save the user's message
    messages.append({
        "role": "user",
        "content": user_input
    })

    # Send the complete conversation to Ollama
    response = ollama.chat(
        model="llama3.2:3b",
        messages=messages
    )

    # Get the model's answer
    reply = response["message"]["content"]

    # Save the model's answer
    messages.append({
        "role": "assistant",
        "content": reply
    })

    return reply
```

```python
def ask_agent(question):

    messages = [
        {
            "role": "user",
            "content": question
        }
    ]

    response = ollama.chat(
        model="llama3.2:3b",
        messages=messages,
        tools=tools
    )

    message = response["message"]

    if message.get("tool_calls"):

        messages.append(message)

        for tool_call in message["tool_calls"]:

            function_name = tool_call["function"]["name"]
            arguments = tool_call["function"]["arguments"]

            if function_name == "read_student_file":

                result = read_student_file(
                    n=arguments["n"]
                )

                messages.append({
                    "role": "tool",
                    "content": result
                })

        final_response = ollama.chat(
            model="llama3.2:3b",
            messages=messages,
            tools=tools
        )

        return final_response["message"]["content"]

    return message["content"]
```

```python
import pandas as pd
import matplotlib.pyplot as plt


def create_student_scatter(
    n: int,
    x_column: str,
    y_column: str
) -> str:
    """
    Read the first N rows from the student dataframe, display those rows,
    and create a scatter plot using two requested columns.

    Use this tool when the user asks for a scatter plot from the student CSV.

    Args:
        n:
            Number of rows to read from the beginning of the CSV.
            For example, 3 means the first 3 rows.

        x_column:
            Exact dataframe column name to place on the X-axis.
            For example: id, math_score, history_score.

        y_column:
            Exact dataframe column name to place on the Y-axis.
            For example: math_score, chemistry_score, history_score.

    Returns:
        A confirmation describing the selected rows and plot axes.
    """

    # Make sure n is a valid integer
    try:
        n = int(n)
    except (TypeError, ValueError):
        return "The number of rows must be an integer."

    # Keep n inside the dataframe size
    n = max(1, min(n, len(df)))

    # Check that the requested columns exist
    if x_column not in df.columns:
        return (
            f"X-axis column '{x_column}' was not found. "
            f"Available columns: {df.columns.tolist()}"
        )

    if y_column not in df.columns:
        return (
            f"Y-axis column '{y_column}' was not found. "
            f"Available columns: {df.columns.tolist()}"
        )

    # Read only the requested number of rows
    selected_rows = df.head(n).copy()

    # Keep the requested columns for plotting
    plot_data = selected_rows[[x_column, y_column]].copy()

    # Convert both axes to numeric data
    plot_data[x_column] = pd.to_numeric(
        plot_data[x_column],
        errors="coerce"
    )

    plot_data[y_column] = pd.to_numeric(
        plot_data[y_column],
        errors="coerce"
    )

    # Remove rows with invalid numeric values
    plot_data = plot_data.dropna()

    if plot_data.empty:
        return (
            f"The columns '{x_column}' and '{y_column}' "
            "do not contain valid numeric data for a scatter plot."
        )

    # Display all requested rows
    print(f"First {n} rows:")
    display(selected_rows)

    # Create the scatter plot
    plt.figure(figsize=(9, 5))

    plt.scatter(
        plot_data[x_column],
        plot_data[y_column]
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(f"{y_column} vs {x_column}")

    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return (
        f"Displayed the first {n} rows and created a scatter plot "
        f"with '{x_column}' on the X-axis and "
        f"'{y_column}' on the Y-axis."
    )
```

```python
def create_student_scatter(
    x_column: str,
    y_column: str
) -> str:
    """
    Display the first N rows from the student CSV and create a
    scatter plot using the requested X-axis and Y-axis columns.

    Use this tool when the user asks for a scatter plot.

    Args:
        x_column:
            Exact dataframe column name to use on the X-axis.

        y_column:
            Exact dataframe column name to use on the Y-axis.

    Returns:
        A confirmation describing the displayed rows and plot axes.
    """

    if x_column not in df.columns:
        return (
            f"Column '{x_column}' was not found. "
            f"Available columns: {df.columns.tolist()}"
        )

    if y_column not in df.columns:
        return (
            f"Column '{y_column}' was not found. "
            f"Available columns: {df.columns.tolist()}"
        )

    selected_rows = df.copy()

    plot_data = selected_rows[[x_column, y_column]].copy()

    plot_data[x_column] = pd.to_numeric(
        plot_data[x_column],
        errors="coerce"
    )

    plot_data[y_column] = pd.to_numeric(
        plot_data[y_column],
        errors="coerce"
    )

    plot_data = plot_data.dropna()

    if plot_data.empty:
        return (
            f"No valid numeric data was found in "
            f"'{x_column}' and '{y_column}'."
        )

    print(f"First {len(selected_rows)} rows:")
    display(selected_rows)

    plt.figure(figsize=(9, 5))

    plt.scatter(
        plot_data[x_column],
        plot_data[y_column]
    )

    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.title(f"{y_column} vs {x_column}")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return (
        f"Inform the user about a scatter plot with '{x_column}' as X and "
        f"'{y_column}' as Y has been created successfully."
    )
```

```python
def ask_agent(question: str) -> str:

    messages = [
        {
            "role": "system",
            "content": f"""
You are a student CSV assistant.

Available dataframe columns:
{df.columns.tolist()}

Available tools:

1. read_student_file
   Use this when the user asks only to display rows.

2. create_student_scatter
   Use this when the user asks for a scatter plot.
   This tool also displays the requested rows.

Rules:
- Extract the requested number of rows.
- Use exact dataframe column names.
- Convert phrases such as "math score" into "math_score".
- If the user asks for rows and a plot, call only
  create_student_scatter.
- Do not invent unavailable columns.
"""
        },
        {
            "role": "user",
            "content": question
        }
    ]

    response = ollama.chat(
        model="llama3.2:3b",
        messages=messages,
        tools=tools
    )

    message = response["message"]
    messages.append(message)

    tool_calls = message.get("tool_calls", [])

    if not tool_calls:
        return message.get("content", "No response returned.")

    results = []

    for tool_call in tool_calls:

        function_name = tool_call["function"]["name"]
        arguments = tool_call["function"]["arguments"]

        if function_name == "read_student_file":

            result = read_student_file(
                n=arguments.get("n", 10)
            )

        elif function_name == "create_student_scatter":

            result = create_student_scatter(
                x_column=arguments.get("x_column", "id"),
                y_column=arguments.get("y_column", "math_score")
            )

        else:
            result = f"Unknown tool: {function_name}"

        results.append(result)

        messages.append({
            "role": "tool",
            "tool_name": function_name,
            "content": str(result)
        })

    final_response = ollama.chat(
        model="llama3.2:3b",
        messages=messages,
        tools=tools
    )

    final_text = final_response["message"].get("content", "")

    if final_text:
        return final_text

    return "\n".join(results)
```
